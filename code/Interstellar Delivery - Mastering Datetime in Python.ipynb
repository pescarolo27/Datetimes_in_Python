# Q1
Define a function called `format_date()`, which formats a timestamp into a readable datetime string.
    - It must accept **two** parameters: `timestamp` - the Unix timestamp integer, and `datetime_format` - a string specifying the desired date format.
    - The function should return the date correctly formatted as a **string**.
    - For example, calling `format_date(1514665153, "%d-%m-%Y")` should output `"30-12-2017"`.

```python
def format_date(timestamp, datetime_format):
    '''
    The function formats a timestamp into a readable datetime string.
    
    Arguments:
        timestamp:        the Unix timestamp integer
        datetime_format:  a string specifying the desired date format
    
    Returns:
        The function returns a date formatted as a string. As an example, the code "format_date(1514665153, "%d-%m-%Y")" 
        should return: "30-12-2017.""
    '''
    
    #To read a Unix timestamp, use datetime.fromtimestamp(timestamp); returns a datetime. Convert Unix timestamp to a datetime.
    d = datetime.fromtimestamp(timestamp)
    
    #Format the datetime as specified
    result = d.strftime(datetime_format)
    
    #Return result
    return result
```

```python
#CHECK -- GOOD
print(type(format_date(1514665153, "%d-%m-%Y")))
print(format_date(1514665153, "%d-%m-%Y"))
```

# Q2
Define a function called `calculate_landing_time()`, which calculates the estimated landing time.
    - It must accept **two** parameters: `rocket_launch_dt` - the rocket launch datetime object, and `travel_duration` - the expected travel time in days as an integer.
    - The function should return the estimated Mars landing time as a datetime **string** in the format `DD-MM-YYYY`.
    - For example, calling `calculate_landing_time(datetime(2023, 2, 15), 20)` should output `"07-03-2023"`.

```python
def calculate_landing_time(rocket_launch_dt, travel_duration):
    '''
    The function calculates the estimated landing time of a rocket.
    
    Arguments:
        rocket_launch_dt:   the rocket launch datetime object.
        travel_duration:    the expected travel time in days as an integer.
    
    Returns:
        The function returns the estimated Mars landing time as a datetime string in the format "DD-MM-YYYY". As an example, 
        calling calculate_landing_time(datetime(2023, 2, 15), 20) should output "07-03-2023".
    '''
    
    #Calculate the estimated landing time using time deltas
    eta = rocket_launch_dt + timedelta(days=travel_duration)
    
    #Format the estimated landing time
    result = eta.strftime('%d-%m-%Y')
    
    #Return result
    return result
```

```python
#CHECK -- GOOD
print(type(calculate_landing_time(datetime(2023, 2, 15), 20)))
print(calculate_landing_time(datetime(2023, 2, 15), 20))
```

Define a function `named days_until_delivery()`, which calculates the days until a package arrives for customers.
    - It must accept **two** parameters: `expected_delivery_dt` - the estimated delivery date as a datetime object for the package, and `current_dt` - the current date as a datetime object.
    - The function should calculate the difference in days between the expected delivery datetime and the current datetime, then return the number of days remaining as an **integer**.
    - For example, calling `days_until_delivery(datetime(2023, 2, 15), datetime(2023, 2, 5))` should output `10`.

```python
def days_until_delivery(expected_delivery_dt, current_dt):
    '''
    The function calculates the days until a package arrives for customers.
    
    Arguments:
        expected_delivery_dt:   the estimated delivery date for the package as a datetime object.
        current_df:             the current date as a datetime object.
    
    Returns:
        The function calculates the difference in days between the expected deliver datetime & the current datetime. 
        It returns the number of days remaining until delivery as an integer. As an example, calling 
        days_until_delivery(datetime(2023, 2, 15), datetime(2023, 2, 5)) should output 10.
    '''
    
    #Calculate time delta between the two specified dates in days.
    result = (expected_delivery_dt - current_dt).days
    
    #Return the result
    return result
```
